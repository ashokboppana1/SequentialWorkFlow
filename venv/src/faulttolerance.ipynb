```python
class CrashState(TypedDict):
    input:str
    step1:str
    step2:str
```

```python
def step_1(state:CrashState):
    print('step1 execution')
    return {'step1' :"done",'input': state['input']}
```

```python
def step_2(state:CrashState):
    print("step2 hanging.. now interrupt")
    time.sleep(20)
    return { 'step2':"done"}
```

```python
def step_3(state:CrashState):
    print("step3 is executing")
    return {"done":True}
```
