```python
from langgraph.graph import START,StateGraph,END
from typing import TypedDict
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
load_dotenv()
```

```python
model=ChatOpenAI()
```

```python
class Blog(TypedDict):
    title:str
    outline:str
    content:str
```

```python
def create_outline(state:Blog) -> Blog:
    title=state['title']

    prompt=f"generate the detailed outline for ablog on the topic -{title}"
    outline=model.invoke(prompt).content

    state['outline']=outline

    return state
```

```python
def create_blog(state:Blog) ->Blog:
    outline=state['outline']
    promppt=f"generate the blog on the fallowing outline {outline}"
    blogcontent=model.invoke(promppt).content
    state['content']=blogcontent
    return state
    
```

```python
chain_graph=StateGraph(Blog)
chain_graph.add_node('generateoutline',create_outline)
chain_graph.add_node('generateblog',create_blog)
chain_graph.add_edge(START,'generateoutline')
chain_graph.add_edge('generateoutline','generateblog')
chain_graph.add_edge('generateblog',END)
```

```python
workflow=chain_graph.compile()
```

```python
blog_outline={'title':'healthy life style'}
result=workflow.invoke(blog_outline)
```
